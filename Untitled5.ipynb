```python
%%writefile Main.java

import java.util.*;

public class Main {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.println("================================");
        System.out.println(" AI FLASHCARD AND QUIZ ENGINE");
        System.out.println("================================");

        String[] questions = {
            "What is Artificial Intelligence?",
            "What is Machine Learning?",
            "What is Deep Learning?",
            "What is NLP?",
            "What is Data Science?"
        };

        String[] answers = {
            "AI enables machines to perform intelligent tasks.",
            "ML allows computers to learn from data.",
            "Deep Learning uses neural networks.",
            "NLP means Natural Language Processing.",
            "Data Science extracts useful information from data."
        };

        System.out.println("\n1. Generate Flashcards");
        System.out.println("2. Start Quiz");

        System.out.print("Enter your choice: ");
        int choice = sc.nextInt();

        if (choice == 1) {
            for (int i = 0; i < questions.length; i++) {
                System.out.println("\nFlashcard " + (i + 1));
                System.out.println("Q: " + questions[i]);
                System.out.println("A: " + answers[i]);
            }
        }
        else if (choice == 2) {
            int score = 0;
            char[] correct = {'A','A','A','A','A'};

            for (int i = 0; i < questions.length; i++) {
                System.out.println("\nQ" + (i+1) + ": " + questions[i]);
                System.out.println("A. Correct concept");
                System.out.println("B. Wrong concept");
                System.out.print("Your answer: ");

                char ans = Character.toUpperCase(sc.next().charAt(0));

                if (ans == correct[i]) {
                    System.out.println("Correct!");
                    score++;
                } else {
                    System.out.println("Wrong!");
                }
            }

            System.out.println("\n===== QUIZ RESULT =====");
            System.out.println("Score: " + score + "/5");
        }
        else {
            System.out.println("Invalid choice!");
        }

        sc.close();
    }
}
```
